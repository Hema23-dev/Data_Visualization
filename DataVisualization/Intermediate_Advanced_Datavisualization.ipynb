```python
# Import necessary libraries

import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
mtcars_data = pd.read_csv("mtcars.csv")
mtcars_data
```

```python
mtcars_data.head(10)
```

# 1. univariate analysis

** Important charts **

* histogram

* countplot

* piechart



## 1.1 Histogram

If we want to know th distribution of one continuous data , we will be using histogram

```python
## HISTOGRAM - use continuous data 

plt.hist(mtcars_data.mpg)
plt.title("mpg distribution")
plt.xlabel("mpg")
plt.ylabel("frequency")
plt.show()
```

## 1.2 COUNTPLOT

 If we want to know th distribution of one discrete data , we will be using countplot

** note **

1. matplotlib doesn't have charts which seaborn has
2. seaborn doesn't have charts which matplotlib has
3. Matplot and Seaborn will have some same charts

```python
sns.countplot(data=mtcars_data,x="am")
plt.title("automatic vs manual car distribution")
plt.show()
```

## 1.3 piechart 

when we want the distribution of one discrete data with value counts

value and respective counts of the column value should be known 

```python
# gives the count for each cylinder value

mtcars_data["cyl"].value_counts()
```

```python
plt.pie(x=mtcars_data["cyl"].value_counts(),labels=["8 cyl","4 cyl","6 cyl"],explode=[0.01,0.01,0.01],autopct="%1.1f%%") 
# labels - the pie division label, explode -> is to seperate each pie by thin white line,autopct = shows percentage for each pie division label
plt.show()
```

# 3. Multivariate analysis

** Important charts **

* pairplot

* heatmap

## 3.1 pairplot

relatonship between one column vs with rest of the columns. And only its takes numeric value of the column

```python
sns.pairplot(data=mtcars_data)
plt.show()
```

## correlation

to identify the strength between two features (can be discrete and continous) and also the dataset should only have numeric column

Pearson Correlation, Coefficient -> -1 to +1 , -1 means negatively correlated , +1 positively correlated



```python
mtcars_data_numeric = mtcars_data.select_dtypes(include="number")
mtcars_data_numeric
```

```python
mtcars_Corr_matrix= mtcars_data_numeric.corr().round(2)   # default correlation is pearson
mtcars_Corr_matrix
```

## 3.2 Heatmap

only for numeric column and we should calculate correlation before the heatmap

```python
sns.heatmap(mtcars_Corr_matrix,annot=True)  # annot = True will give the value inside the matrix
plt.show()
```

## NOTES:

0 - 0.20 = VERY WEAK CORRELATION

0.21 - 0.40 = WEAK CORRELATION

0.41 - 0.60 = NORMAL CORRELATION OR STANDARD CORRELATION

0.61 - 0.80 = STRONG CORRELATION

0.81 - 1.00 = VERY STRONG CORRELATION

For example : Here the mpg (-0.85) is very strongly negatively correlated with the cyl(cylinders) which means the number of cyl increases the mpg of the car decreases.

## Advanced Data Visualization


Plotly is used when we need the charts to be interactive .

Here we will get charts in dynamic way , it can be changed when we click the graphs

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'Other countries' # Represent only large countries
fig = px.pie(df, values='pop', names='country', title='Population of European continent')
fig.show()
```

## Notes:

Here in plotly ,

1. Univariate charts can include more than one feature - > x , hue(another feature)

2. Bivariate charts can include more than two feature
